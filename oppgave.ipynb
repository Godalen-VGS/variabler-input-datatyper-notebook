## Variabler

Variabler i programmering fungerer som bokser eller beholdere som holder på informasjon. Denne informasjonen kan være tekst (strenger), tall (heltall eller desimaltall), lister og mange andre datatyper. I Python kan du opprette en variabel ved å gi den et navn og deretter tildele den en verdi med et likhetstegn (`=`).

### Eksempler

Tilordning av et heltall til en variabel:
```python
alder = 18
```
Tilordning av en streng til en variabel:
```python
navn = "Ola"
```
Tilordning av et desimaltall (float) til en variabel:
```python
hoyde = 1.75  # Høyde i meter
```

## Datatyper

I programmering refererer datatyper til de ulike typene verdier som en variabel kan holde. Noen av de mest vanlige datatypene i Python er:

**Integers (heltall)**: positive eller negative hele tall, uten desimalpunkt. F.eks. 1, -5, 0, 1000.
```python
mitt_heltall = 10
```

**Float (desimaltall)**: tall med et desimalpunkt. F.eks. 3.14, -0.01, 2.0.
```python
pi = 3.14
```

**String (tekst)**: sekvenser av karakterer, innenfor enkle eller doble anførselstegn. F.eks. "Hei", 'Python'.
```python
tekst = "Python er gøy!"
```

**Boolean**: enten `True` eller `False`. Brukes ofte i betingelser og logiske operasjoner.
```python
roboter_er_kult = True
```

**Lister**: en samling av verdier, innenfor firkantparanteser. F.eks. `[1, 2, 3]` eller `["a", "b", "c"]`.
```python
karakterer = [1, 2, 3, 4, 5, 6]
```

## Input

I Python kan vi bruke den innebygde funksjonen `input()` for å hente informasjon fra brukeren. Når denne funksjonen kjøres, stopper programmet opp og venter på at brukeren skal skrive noe. Det brukeren skriver returneres som en streng.
```python
bruker_navn = input("Hva er navnet ditt? ")
print(f"Hei, {bruker_navn}!")
```

## Typekonvertering i Python

Ofte trenger vi å konvertere en verdi fra én datatype til en annen — dette kalles typekonvertering.

Fra streng til heltall/desimaltall:
```python
streng_til_int = int("123")
streng_til_float = float("123.45")
```
Fra heltall/desimaltall til streng:
```python
int_til_streng = str(123)
float_til_streng = str(123.45)
```
Fra heltall til desimaltall og omvendt:
```python
int_til_float = float(123)
float_til_int = int(123.45)  # Dette vil avrunde tallet ned
```
Fra andre datatyper til boolske verdier: nesten alle verdier blir `True`, unntatt "tomme"/"null"-verdier som `0`, `""`, `[]` og `None`, som blir `False`.
```python
bool_av_int = bool(1)        # True
bool_av_tom_streng = bool("")  # False
```
